```python
import pandas as pd

pd.set_option("display.max_columns", None)
RAW = "../data/raw/"
```

```python
files = ["claims", "claims_transactions", "payers", "patients", "providers", "organizations", "encounters"]

dfs = {}

for name in files:
    dfs[name] = pd.read_csv(RAW + f"{name}.csv", low_memory=False)
    print(f"--- {name}: {dfs[name].shape} ---")
    print(list(dfs[name].columns))
    print()
```

```python
c = dfs["claims"]

for col in ["STATUS1", "STATUS2", "STATUSP"]:
    print(c[col].value_counts(dropna=False), "\n")
```

```python
c["SERVICEDATE"] = pd.to_datetime(c["SERVICEDATE"])

print("Service date:", c["SERVICEDATE"].min(), "->", c["SERVICEDATE"].max())

print(c[["OUTSTANDING1", "OUTSTANDING2", "OUTSTANDINGP"]].describe())
```
